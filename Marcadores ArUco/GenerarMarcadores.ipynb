```python
import cv2
import numpy as np
import os


# PARÁMETROS
num_markers = 8            # cantidad total de marcadores
marker_size_cm = 8         # tamaño del marcador (sin borde) en cm
borde_cm = 1               # borde blanco alrededor del marcador en cm
dpi = 400                  # resolución de impresión
cols = 1                   # columnas por hoja (1 por columna, 2 en total)
espacio_cm = 2             # espacio entre marcadores en cm
markers_per_sheet = 2      # marcadores por hoja

# CONVERSIÓN DE CM A PÍXELES
cm_to_px = dpi / 2.54
marker_size_px = int(marker_size_cm * cm_to_px)
borde_px = int(borde_cm * cm_to_px)
espacio_px = int(espacio_cm * cm_to_px)

# Diccionario ArUco
aruco_dict = cv2.aruco.getPredefinedDictionary(cv2.aruco.DICT_4X4_50)

# Tamaño A4 en píxeles
a4_width_px = int(21.0 * cm_to_px)
a4_height_px = int(29.7 * cm_to_px)

# Función para agregar borde blanco
def agregar_borde(marker_img, borde):
    return cv2.copyMakeBorder(marker_img, borde, borde, borde, borde,
                              borderType=cv2.BORDER_CONSTANT, value=255)

# GENERAR HOJAS A4 CON LOS MARCADORES
for sheet_idx in range(int(np.ceil(num_markers / markers_per_sheet))):
    sheet = np.ones((a4_height_px, a4_width_px), dtype=np.uint8) * 255
    start_x = int((a4_width_px - marker_size_px - 2 * borde_px) / 2)  # centrar horizontalmente
    start_y = int(3 * cm_to_px)  # margen superior

    for i in range(markers_per_sheet):
        marker_id = sheet_idx * markers_per_sheet + i
        if marker_id >= num_markers:
            break

        marker = cv2.aruco.generateImageMarker(aruco_dict, marker_id, marker_size_px)
        marker_bordered = agregar_borde(marker, borde_px)

        # Posición vertical: uno arriba, otro más abajo
        y = start_y + i * (marker_bordered.shape[0] + espacio_px)
        x = start_x

        sheet[y:y+marker_bordered.shape[0], x:x+marker_bordered.shape[1]] = marker_bordered

        # Dibujar un recuadro fino (2 px) alrededor del marcador
        cv2.rectangle(sheet, (x, y),
                      (x + marker_bordered.shape[1], y + marker_bordered.shape[0]),
                      color=0, thickness=2)  # negro, 2 px de grosor

        # Agregar ID debajo del marcador
        cv2.putText(sheet, f"ID {marker_id}",
                    (x + marker_bordered.shape[1]//3, y + marker_bordered.shape[0] + int(0.6*cm_to_px)),
                    cv2.FONT_HERSHEY_SIMPLEX, 1, 0, 2, cv2.LINE_AA)

    # Guardar hoja
    sheet_path = os.path.join(output_dir, f"marcadores_A4_hoja_{sheet_idx+1}.png")
    cv2.imwrite(sheet_path, sheet)
    print(f"✅ Hoja {sheet_idx+1} guardada en: {sheet_path}")
```
